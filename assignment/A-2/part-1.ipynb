```python
import tokenizers, transformers
from transformers import AutoTokenizer, GPT2TokenizerFast

print(tokenizers.__version__)

tokenizer = GPT2TokenizerFast.from_pretrained('gpt2')

print('Class:', type(tokenizer).__name__)
print('is_fast:', tokenizer.is_fast)
print('eos_token:', tokenizer.eos_token, '| id:', tokenizer.eos_token_id)
print('bos_token:', tokenizer.bos_token, '| id:', tokenizer.bos_token_id)
print('vocab_size:', tokenizer.vocab_size)
# Quick test
sample = 'Hello world this is a test.'
ids = tokenizer(sample, add_special_tokens=False)['input_ids']
print('Tokens:', ids, '→ with EOS:', ids + [tokenizer.eos_token_id])
```
